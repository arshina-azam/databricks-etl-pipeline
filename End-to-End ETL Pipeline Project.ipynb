```python
%python
data = [
    (1, " Arshina ", "IT", 30000),
    (2, "Rahul", "HR", 25000),
    (3, "Priya ", "IT", 35000),
    (2, "Rahul", "HR", 25000),
    (4, "Aman ", "Finance", 40000)
]

columns = ["id", "name", "department", "salary"]

df = spark.createDataFrame(data, columns)

display(df)
```

```python
%python
from pyspark.sql.functions import trim, col

clean_df = df.dropDuplicates()

clean_df = clean_df.withColumn(
    "name",
    trim(col("name"))
)

clean_df = clean_df.withColumn(
    "department",
    trim(col("department"))
)

display(clean_df)
```

```python
%python
clean_df.write.mode("overwrite").saveAsTable("employee_target")
```

```python
data = [
    (101, " Arshina ", "IT", 35000, "Kolkata"),
    (102, "Rahul", "HR", 28000, "Delhi"),
    (103, "Priya ", "IT", 42000, "Mumbai"),
    (103, "Priya ", "IT", 42000, "Mumbai"),
    (104, "Aman", "Finance", 45000, "Pune"),
    (105, "", "HR", 32000, "Delhi"),
    (106, " Sam ", "IT", 38000, "Delhi"),
    (107, "Neha", "Finance", 50000, "Kolkata"),
    (108, "Rohan", "IT", 0, "Mumbai"),
    (109, "Riya", "HR", 30000, "Kolkata")
]

columns = ["employee_id", "name", "department", "salary", "city"]

df_csv = spark.createDataFrame(data, columns)

display(df_csv)
```

```python
from pyspark.sql.functions import col, trim

# Remove exact duplicate records
deduplicated_df = df_csv.dropDuplicates()

# Remove extra spaces from text columns
clean_df = deduplicated_df.withColumn(
    "name", trim(col("name"))
).withColumn(
    "department", trim(col("department"))
).withColumn(
    "city", trim(col("city"))
)

display(clean_df)
```

```python
invalid_df = clean_df.filter(
    col("name").isNull()
    | (trim(col("name")) == "")
    | (col("salary") <= 0)
)

display(invalid_df)
```

```python
from pyspark.sql.functions import col, trim

# Remove records with missing names or invalid salaries
silver_df = clean_df.filter(
    (col("name").isNotNull()) &
    (trim(col("name")) != "") &
    (col("salary") > 0)
)

display(silver_df)
```

```python
silver_df.write.mode("overwrite").saveAsTable("employees_silver")

print("Silver Layer created successfully!")
```

```python
from pyspark.sql.functions import count, sum, avg, round, col

# Department-wise salary analysis
gold_df = silver_df.groupBy("department").agg(
    count("*").alias("total_employees"),
    sum("salary").alias("total_salary"),
    round(avg("salary"), 2).alias("average_salary")
)

display(gold_df)
```

```python
gold_df.write.mode("overwrite").saveAsTable("employees_gold")

print("Gold Layer created successfully!")
```
