```python
%python
data = [
    (1, " Arshina ", "IT", 30000),
    (2, "Rahul", "HR", 25000),
    (3, "Priya ", "IT", 35000),
    (2, "Rahul", "HR", 25000),
    (4, "Aman ", "Finance", 40000)
]

columns = ["id", "name", "department", "salary"]

df = spark.createDataFrame(data, columns)

display(df)
```

```python
%python
from pyspark.sql.functions import trim, col

clean_df = df.dropDuplicates()

clean_df = clean_df.withColumn(
    "name",
    trim(col("name"))
)

clean_df = clean_df.withColumn(
    "department",
    trim(col("department"))
)

display(clean_df)
```

```python
%python
clean_df.write.mode("overwrite").saveAsTable("employee_target")
```

```python
%python
result_df = spark.table("employee_target")

display(result_df)
```

```python
%python
# Count source records
source_count = df.count()

# Count cleaned records
target_count = clean_df.count()

print("Source Record Count:", source_count)
print("Target Record Count:", target_count)

print("Records Removed:", source_count - target_count)
```

```python
%python
clean_df.createOrReplaceTempView("employee_view")
```

```python
%sql
SELECT * FROM employee_view;
```

```python
%sql
SELECT * FROM employee_view;
```

```python
%sql
SELECT
    department,
    COUNT(*) AS total_employees,
    SUM(salary) AS total_salary,
    AVG(salary) AS average_salary
FROM employee_view
GROUP BY department;
```

```python
%sql
SELECT *
FROM employee_view
WHERE salary <= 0
   OR name IS NULL
   OR department IS NULL;
```

```python
%sql
SELECT name, department, salary
FROM employee_view
ORDER BY salary DESC;
```

```python
# Save original data in Bronze Layer
df.write.mode("overwrite").saveAsTable("employee_bronze")

print("Bronze Layer created successfully!")
```

```python
# Save cleaned data in Silver Layer
clean_df.write.mode("overwrite").saveAsTable("employee_silver")

print("Silver Layer created successfully!")
```

```python
from pyspark.sql.functions import count, sum, avg

gold_df = clean_df.groupBy("department").agg(
    count("*").alias("total_employees"),
    sum("salary").alias("total_salary"),
    avg("salary").alias("average_salary")
)

gold_df.write.mode("overwrite").saveAsTable("employee_gold")

display(gold_df)
```
